```python
#Descriptive analysis
import statistics
import matplotlib.pyplot as plt
mem_names=["Amrutha","anu","Thanu","asritha","Akhila"]
ages=[22,25,24,23,26]
workout_hours=[2,1.5,2.5,1,3]
calories_burned=[500,250,600,300,750]
print("fitness center dataset")
for i in range(len(mem_names)):
  print(mem_names[i],ages[i],workout_hours[i],calories_burned[i])
avg_workout=sum(workout_hours)/len(workout_hours)
print("average workout hours:",avg_workout)
max_calories=max(calories_burned)
print("maximum calories burned:",max_calories)
min_calories=min(calories_burned)
print("minimum calories burned:",min_calories)
median_calories=statistics.median(calories_burned)
print("median calories burned:",median_calories)
std_dev=statistics.stdev(calories_burned)
print("standard deviation of calories burned:",std_dev)
index=calories_burned.index(max_calories)
print("person with maximum calories burned:",mem_names[index])
index=calories_burned.index(min_calories)
plt.bar(mem_names,calories_burned)
plt.xlabel("member names")
plt.ylabel("calories burned")
plt.title("Fitness analysis - Bar Chart")
plt.show()
plt.pie(calories_burned, labels=mem_names)
plt.title("Fitness analysis - Calories Burned Distribution")
plt.axis('equal')
plt.show()
plt.scatter(mem_names,calories_burned)
plt.xlabel("member names")
plt.ylabel("calories burned")
plt.title("Fitness analysis - Scatter Plot")
plt.show()
```

```python
import pandas as pd
df=pd.read_excel("/content/exercise_dataset.xlsx")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import statistics
print("--- Fitness Center Dataset Analysis (from DataFrame) ---")
print("Descriptive statistics for 'Calories Burn':")
print(df['Calories Burn'].describe())
max_calories_val = df['Calories Burn'].max()
min_calories_val = df['Calories Burn'].min()
highest_calories_row = df[df['Calories Burn'] == max_calories_val].iloc[0]
lowest_calories_row = df[df['Calories Burn'] == min_calories_val].iloc[0]
print(f"\nHighest calories burned: {max_calories_val:.2f}")
print(f"Details of person with highest calories burned:")
print(f"  Index: {highest_calories_row.name}, Age: {highest_calories_row['Age']}, Gender: {highest_calories_row['Gender']}")
print(f"\nLowest calories burned: {min_calories_val:.2f}")
print(f"Details of person with lowest calories burned:")
print(f"  Index: {lowest_calories_row.name}, Age: {lowest_calories_row['Age']}, Gender: {lowest_calories_row['Gender']}")
gender_calories = df.groupby('Gender')['Calories Burn'].sum()
plt.figure(figsize=(8, 8))
plt.pie(gender_calories, labels=gender_calories.index)
plt.title('Total Calories Burned by Gender')
plt.axis('equal')
plt.show()
print("\n Most and Least Active Members Analysis")
max_duration_val = df['Duration'].max()
min_duration_val = df['Duration'].min()
highest_duration_row = df[df['Duration'] == max_duration_val].iloc[0]
lowest_duration_row = df[df['Duration'] == min_duration_val].iloc[0]
print(f"\nHighest workout duration: {max_duration_val} minutes")
print(f"Details of person with highest workout duration:")
print(f"  Index: {highest_duration_row.name}, Age: {highest_duration_row['Age']}, Gender: {highest_duration_row['Gender']}")
print(f"\nLowest workout duration: {min_duration_val} minutes")
print(f"Details of person with lowest workout duration:")
print(f"  Index: {lowest_duration_row.name}, Age: {lowest_duration_row['Age']}, Gender: {lowest_duration_row['Gender']}")
df_sorted_duration = df.sort_values(by='Duration', ascending=False)
```
