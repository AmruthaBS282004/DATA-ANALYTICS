```python
import pandas as pd
import numpy as np
marks=[70,80,90,60,100]
hours=[2,3,5,1,6]
print("Mean of marks:", np.mean(marks))
print("Median of marks:", np.median(marks))
print("Variance of marks:", np.var(marks))
print("Standard deviation of marks:", np.std(marks))
print("Correlation between marks and hours:", np.corrcoef(marks, hours)[0, 1])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8, 6))
sns.scatterplot(x=hours, y=marks, s=100, hue=marks, size=hours, legend='full', palette='viridis')
plt.title('Marks vs. Study Hours')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

marks = [70, 80, 90, 60, 100]
hours = [2, 3, 5, 1, 6]

print("Mean of marks:", np.mean(marks))
print("Median of marks:", np.median(marks))
print("Variance of marks:", np.var(marks))
print("Standard deviation of marks:", np.std(marks))
print("Correlation between marks and hours:", np.corrcoef(marks, hours)[0,1])

plt.plot(hours, marks, marker='o')
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

sales = [2000, 2500, 3000, 3500, 4000]
profit = [400, 500, 650, 700, 850]

print("Mean of sales:", np.mean(sales))
print("Median of sales:", np.median(sales))
print("Variance of sales:", np.var(sales))
print("Standard deviation of sales:", np.std(sales))
print("Correlation between sales and profit:", np.corrcoef(sales, profit)[0,1])

plt.bar(sales, profit)
plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

experience = [1, 2, 3, 4, 5]
salary = [25000, 30000, 35000, 42000, 50000]

print("Mean salary:", np.mean(salary))
print("Median salary:", np.median(salary))
print("Variance of salary:", np.var(salary))
print("Standard deviation of salary:", np.std(salary))
print("Correlation between experience and salary:", np.corrcoef(experience, salary)[0,1])

plt.plot(experience, salary, marker='o')
plt.title("Experience vs Salary")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
temperature = [30, 32, 35, 31, 36]

print("Mean temperature:", np.mean(temperature))
print("Median temperature:", np.median(temperature))
print("Variance of temperature:", np.var(temperature))
print("Standard deviation of temperature:", np.std(temperature))
print("Correlation between days and temperature:", np.corrcoef(days, temperature)[0,1])

plt.plot(days, temperature, marker='o')
plt.title("Days vs Temperature")
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.grid(True)
plt.show()
```
