```python
import pandas as pd
import matplotlib.pyplot as plt
data = {
    "Order_Hour": [8, 10, 1, 2, 4, 6, 7, 1],
    "Delivery_Time": [30, 45, 50, 25, 55, 35, 50, 25],
    "Delivery_Partner": ["Amrutha", "Akhila", "Arjun", "Priya", "Kiran", "Meena", "Ravi", "Divya"],
    "Area": ["A", "B", "C", "A", "D", "B", "C", "D"],
    "Customer_Rating": [5, 4, 3, 5, 2, 4, 3, 5]}
df = pd.DataFrame(data)
avg_time = df["Delivery_Time"].mean()
print("Average Delivery Time:", avg_time)
peak_hour = df["Order_Hour"].mode()
print("Peak Order Hour:", peak_hour)
delay_area = df.groupby("Area")["Delivery_Time"].mean()
print("\nMost Delayed Areas:")
print(delay_area)
top_partner = df.groupby("Delivery_Partner")["Customer_Rating"].mean()
print("\nTop Rated Delivery Partners:")
print(top_partner)
delay_area.plot(kind="bar")
plt.title("Delivery Time by Area")
plt.xlabel("Area")
plt.ylabel("Minutes")
plt.show()

plt.scatter(df["Order_Hour"], df["Delivery_Time"])
plt.title("Order Hour vs Delivery Time")
plt.xlabel("Order Hour")
plt.ylabel("Delivery Time")
plt.show()

plt.pie(df["Customer_Rating"].value_counts(), labels=df["Customer_Rating"].unique())
plt.title("Customer Rating Distribution")
plt.xlabel("Customer Rating")
plt.ylabel("Frequency")
plt.show()

print("\nPerformance Report")
print("Average delivery time is", avg_time, "minutes")
print("Peak order hour is", peak_hour)

print("\nRecomendations")
print("1. Increase staff during peak hours")
print("2. Improve delivery in delayed areas")
print("3. Train low-rated delivery partners")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data={
    "product_purchases":["Laptop","Mouse","Keyboard","Laptop","watch"],
    "cart_abondement":[0,1,1,0,1],
    "customer_ratings":[2,5,1,4,3],
    "payment_methods":["upi","Cash on delivery","creditcard","upi","Debitcard"]
}
df=pd.DataFrame(data)
most_soldproducts=df["product_purchases"].mode()
print(f"\nMost Sold Products: {most_soldproducts.to_string(index=False)}")
high_revenue_customers=df["customer_ratings"].max()
print(f"\nHigh Revenue Customers: {high_revenue_customers}")
most_paymentmethod=df["payment_methods"].mode()
print(f"\nMost used payment methods : {most_paymentmethod.to_string(index=False)}")
poor_rating_products=df[df["customer_ratings"]<3]["product_purchases"]
print(f"\nProducts with poor rating: {poor_rating_products.to_string(index=False)}")
plt.figure(figsize=(9, 6))
plt.hist(df["customer_ratings"])
plt.xlabel('Product Purchases')
plt.ylabel('Cart Abondment')
plt.title('\nCart Abondment by Product Purchases')
plt.show()

product_counts = df["product_purchases"].value_counts()
plt.figure(figsize=(9, 6))
plt.pie(product_counts, labels=product_counts.index)
plt.title('\nDistribution of Product Purchases')
plt.axis('equal')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data = {
    "Transaction_Amount": [500.50, 20000.75, 300.25, 45000.90, 700.40],
    "Location": ["India", "USA", "India", "Unknown", "India"],
    "Time": ["10AM", "2AM", "1PM", "3AM", "11AM"],
    "Account_Activity": [5, 50, 3, 80, 4]
}
df = pd.DataFrame(data)
print(df)
unusual = df[df["Transaction_Amount"] > 10000]
print("\nUnusual Transactions:")
print(unusual)
high_risk = df[df["Account_Activity"] > 40]
print("\nHigh Risk Accounts:")
print(high_risk)
fraud = df[
    (df["Transaction_Amount"] > 10000) &
    (df["Time"].isin(["2AM", "3AM"]))
]
print("\nPossible Fraud Patterns:")
print(fraud)
plt.bar(df.index, df["Transaction_Amount"])
plt.title("Transaction Amounts")
plt.xlabel("Transactions")
plt.ylabel("Amount")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data = {
    "Transaction_Amount": [500.50, 20000.00, 300.25, 45000.00, 700.40, 15000.00],
    "Location": ["India", "America", "India", "Unknown", "India", "UK"],
    "Time": ["10AM", "2AM", "1PM", "3AM", "11AM", "12AM"],
    "Account_Activity": [5, 50, 3, 80, 4, 60]
}
df = pd.DataFrame(data)
print(" TRANSACTION DATASET")
print(df)
unusual = df[df["Transaction_Amount"] % 1 == 0]
print("\n UNUSUAL TRANSACTIONS")
print(unusual)
high_risk = unusual
print("\n HIGH RISK ACCOUNTS")
print(high_risk)
fraud = high_risk[
    high_risk["Time"].isin(["12AM", "1AM", "2AM", "3AM"])
]
print("\n POSSIBLE FRAUD PATTERNS ")
print(fraud)
print("\n SUMMARY ")
print("Total Transactions:", len(df))
print("Unusual Transactions:", len(unusual))
print("High Risk Accounts:", len(high_risk))
print("Possible Fraud Cases:", len(fraud))
plt.bar(df.index, df["Transaction_Amount"])
plt.title("Transaction Amount Analysis")
plt.xlabel("Transaction Number")
plt.ylabel("Transaction Amount")
plt.show()
```
